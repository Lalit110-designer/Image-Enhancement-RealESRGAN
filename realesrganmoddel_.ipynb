```python
import sys, types
import torchvision.transforms.functional as F

mod = types.ModuleType("torchvision.transforms.functional_tensor")
mod.rgb_to_grayscale = F.rgb_to_grayscale
sys.modules["torchvision.transforms.functional_tensor"] = mod
```

```python
from realesrgan import RealESRGANer
from basicsr.archs.rrdbnet_arch import RRDBNet
import cv2, urllib.request, os
```

```python
os.makedirs("weights", exist_ok=True)

model_url  = "https://github.com/xinntao/Real-ESRGAN/releases/download/v0.1.0/RealESRGAN_x4plus.pth"
model_path = "weights/RealESRGAN_x4plus.pth"

if not os.path.exists(model_path):
    urllib.request.urlretrieve(model_url, model_path)
    print("Downloaded!")
```

```python
model = RRDBNet(
    num_in_ch=3,   num_out_ch=3,
    num_feat=64,   num_block=23,
    num_grow_ch=32, scale=4
)

upsampler = RealESRGANer(
    scale=4,
    model_path='weights/RealESRGAN_x4plus.pth',
    model=model,
    device='cuda'   # use 'cpu' if no GPU
)
```

```python
# Option A — Colab upload widget
from google.colab import files
uploaded = files.upload()
input_filename = list(uploaded.keys())[0]

# Option B — manual path
# input_filename = 'input.jpg'
```

```python
img = cv2.imread('house_blur_image.jfif', cv2.IMREAD_UNCHANGED)

output, _ = upsampler.enhance(img, outscale=4)

cv2.imwrite('output.jpg', output)
print("Done! Saved to output.jpg")
```
